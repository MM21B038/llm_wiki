```python
from fastmcp import FastMCP
```

```python
import subprocess
```

```python
from pathlib import Path
```

```python
# mcp = FastMCP("MyServer")
```

```python
mcp = FastMCP(
    "FileReader",
    instructions="Provide tools to read a file"
)
```

```python
@mcp.tool
def read_file(path: str) -> str:
    file = Path(path)
    
    if not file.exists():
        return f"File does not exists: {path}"
    
    return file.read_text(encoding="utf-8")
```

```python
print(mcp)
```

```python
@mcp.tool
def execute_command(command: str) -> str:
    result = subprocess.run(
    command,
    shell=True,
    capture_output=True,
    text=True,
    timeout=30,
    )

    return result.stdout
```

```python
tools = await mcp.list_tools()

for tool in tools:
    print(tool.name)
```
