```python
import time

#connect
client = redis.Redis(host='localhost', port=6379, decode_responses=True)

# Define our queue name
QUEUE_NAME = 'my_task_queue'

# --- PRODUCER: Add tasks to the queue ---
def push_task(task_data):
    # LPUSH adds the item to the left (head) of the list
    client.lpush(QUEUE_NAME, task_data)
    print(f"[-] Enqueued: {task_data}")

# --- CONSUMER: Process tasks from the queue ---
def process_tasks():
    print("[*] Starting queue listener. Press Stop in Jupyter to exit.")
    
    while True:
        # BRPOP is a blocking pop. It waits up to 2 seconds for an item.
        # It returns a tuple: (queue_name, item_value)
        result = client.brpop(QUEUE_NAME, timeout=2)
        
        if result:
            _, task = result
            print(f"[+] Processing task: {task}")
            
            # Simulate some work
            time.sleep(1)
            print(f"[✔] Completed: {task}")
        else:
            print("[.] Queue is empty, waiting...")
            break # Exit loop if empty for demo purposes (remove 'break' for a persistent worker)
```

```python
push_task("Task 1: Send Email")
push_task("Task 2: Process Image")
push_task("Task 3: Generate Report")
```

```python
process_tasks()
```
